```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from src.data_loader import DataLoader
from src.chunker import RecursiveChunker
from src.vectorstore import VectorStore
from src.rag_pipeline import RAGPipeline
print('All modules loaded successfully.')
```

```python
loader = DataLoader()
docs = loader.load_directory('../data/raw_documents')
print(f'Loaded {len(docs)} document pages.')
chunker = RecursiveChunker(chunk_size=500, chunk_overlap=80)
chunks = chunker.chunk_documents(docs)
print(f'Generated {len(chunks)} text chunks.')
vstore = VectorStore(score_threshold=0.15)
vstore.add_chunks(chunks)
print(f'VectorStore indexed {len(vstore.chunks)} chunks.')
```

```python
pipeline = RAGPipeline(vectorstore=vstore, disclaimer='"' + project_meta['disclaimer'] + '"')
response = pipeline.query('What are the key policy requirements?')
print('Query Answer:\n', response.answer)
print('\nCitations:\n', response.citations)
```
